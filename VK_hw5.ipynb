```python
from collections import deque
import heapq

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def build_tree(arr, i=0):
    if i >= len(arr) or arr[i] is None:
        return None
    root = TreeNode(arr[i])
    root.left = build_tree(arr, 2 * i + 1)
    root.right = build_tree(arr, 2 * i + 2)
    return root
```

# Задача 1. Полное бинарное дерево

## Условие
Реализовать функцию проверки, является ли дерево полным. Полное бинарное дерево — это дерево, в котором все уровни, кроме, возможно, последнего, полностью заполнены, а на последнем уровне все узлы максимально смещены влево.

**Что нужно сделать:**
Корректно определить условие заполненности уровней и объяснить выбранный способ обхода (например, BFS).

```python
def is_complete_tree(root):
    if not root:
        return True

    queue = deque([root])
    seen_missing = False

    while queue:
        node = queue.popleft()
        if node is None:
            seen_missing = True
        else:
            if seen_missing:
                return False
            queue.append(node.left)
            queue.append(node.right)

    return True

raw = input("Введите элементы массива через пробел (None для пустых): ").split()
arr = []
for x in raw:
    if x == 'None':
        arr.append(None)
    else:
        arr.append(int(x))
# -------------------------

root = build_tree(arr)
result = is_complete_tree(root)
print(f"Является ли дерево полным: {result}")
```

Используем обход в ширину (BFS) через очередь. Идея: когда мы встречаем первый узел с отсутствующим ребенком (или пустую ячейку), все последующие узлы в очереди также не должны иметь детей. Если после этого мы встречаем узел с детьми — дерево не полное.

# Задача 2. Объединение K отсортированных массивов

## Условие
Напишите функцию, объединяющую \(K\) отсортированных массивов в один отсортированный массив. Для эффективного решения используйте мин-кучу и проанализируйте сложность алгоритма.

```python
def merge_k_sorted(arrays):
    result = []
    heap = []

    for i, arr in enumerate(arrays):
        if arr:
            heapq.heappush(heap, (arr[0], i, 0))

    while heap:
        val, arr_idx, elem_idx = heapq.heappop(heap)
        result.append(val)

        if elem_idx + 1 < len(arrays[arr_idx]):
            heapq.heappush(heap, (arrays[arr_idx][elem_idx + 1], arr_idx, elem_idx + 1))

    return result

k = int(input("Введите количество массивов: "))
arrays = []
for i in range(k):
    arr = list(map(int, input(f"Введите элементы массива {i+1} через пробел: ").split()))
    arrays.append(arr)

result = merge_k_sorted(arrays)
print(f"Объединенный массив: {result}")
```

Сложность: O(N*log K), где N — общее число элементов во всех массивах, K — количество массивов. В куче всегда не более K элементов, поэтому операции за log K.

# Задача 3. K-ый наименьший / наибольший элемент в массиве

## Условие
Дан массив чисел. Необходимо найти K-ый наименьший или K-ый наибольший элемент. Рассмотрите эффективные подходы (например, с использованием кучи) и обоснуйте сложность решения.

```python
def kth_smallest(nums, k):
    max_heap = []
    for num in nums:
        heapq.heappush(max_heap, -num)
        if len(max_heap) > k:
            heapq.heappop(max_heap)
    return -max_heap[0]

def kth_largest(nums, k):
    min_heap = []
    for num in nums:
        heapq.heappush(min_heap, num)
        if len(min_heap) > k:
            heapq.heappop(min_heap)
    return min_heap[0]

nums = list(map(int, input("Введите элементы массива через пробел: ").split()))
k = int(input("Введите K: "))

print(f"{k}-ый наименьший элемент: {kth_smallest(nums, k)}")
print(f"{k}-ый наибольший элемент: {kth_largest(nums, k)}")
```

Сложность: O(n*log K), где n — длина массива, K — искомый номер.
Мы проходим по массиву один раз. Для каждого элемента делаем heappush в кучу размера K (за O(log K)). Если куча переполняется, делаем heappop (тоже за O(log K))

# Задача 4. K-ый наименьший / наибольший элемент в BST

## Условие
Дано бинарное дерево поиска. Реализуйте алгоритм нахождения K-ого наименьшего или наибольшего элемента, используя свойства BST и обход дерева.

```python
def kth_smallest_bst(root, k):
    stack = []
    curr = root
    count = 0
    while stack or curr:
        while curr:
            stack.append(curr)
            curr = curr.left
        curr = stack.pop()
        count += 1
        if count == k:
            return curr.val
        curr = curr.right
    return None

def kth_largest_bst(root, k):
    stack = []
    curr = root
    count = 0
    while stack or curr:
        while curr:
            stack.append(curr)
            curr = curr.right
        curr = stack.pop()
        count += 1
        if count == k:
            return curr.val
        curr = curr.left
    return None

arr = list(map(int, input("Введите элементы BST через пробел: ").split()))
k = int(input("Введите K: "))

root = build_tree(arr)
print(f"{k}-ый наименьший в BST: {kth_smallest_bst(root, k)}")
print(f"{k}-ый наибольший в BST: {kth_largest_bst(root, k)}")
```

# Задача 5. Balance factor

## Условие
На вход функции передаётся указатель на корень бинарного дерева поиска. Необходимо для каждого узла вычислить и заполнить поле balance_factor (разницу высот левого и правого поддеревьев), объяснив способ вычисления высоты.

```python
def height(node):
    if not node:
        return 0
    return 1 + max(height(node.left), height(node.right))

def fill_balance_factors(root):
    if not root:
        return
    root.balance_factor = height(root.left) - height(root.right)
    fill_balance_factors(root.left)
    fill_balance_factors(root.right)

def print_balance_factors(root):
    if not root:
        return
    print(f"Узел {root.val}: balance_factor = {root.balance_factor}")
    print_balance_factors(root.left)
    print_balance_factors(root.right)

raw = input("Введите элементы BST через пробел (None для пустых): ").split()
arr = [None if x == 'None' else int(x) for x in raw]

root = build_tree(arr)

fill_balance_factors(root)
print("\nBalance factors для каждого узла:")
print_balance_factors(root)
```

Высота узла — это количество ребер на longest пути от узла до листа. Вычисляется рекурсивно: height(node) = 1 + max(height(left), height(right)).

Для пустого узла высота равна 0.

balance_factor для каждого узла — это разница высот его левого и правого поддеревьев: height(left) - height(right).

# Задача 6. Преобразование в зеркальное дерево

## Условие
Реализуйте алгоритм зеркального отражения бинарного дерева, меняя местами левые и правые поддеревья каждого узла. Поясните, почему алгоритм корректен и какова его временная сложность.

```python
def mirror_tree(root):
    if not root:
        return None
    root.left, root.right = root.right, root.left
    mirror_tree(root.left)
    mirror_tree(root.right)
    return root

def print_in_order(root):
    if not root:
        return []
    result = []
    result.extend(print_in_order(root.left))
    result.append(root.val)
    result.extend(print_in_order(root.right))
    return result

arr = list(map(int, input("Введите элементы дерева через пробел: ").split()))
root = build_tree(arr)

print(f"Исходное дерево (in-order): {print_in_order(root)}")
mirror_tree(root)
print(f"Зеркальное дерево (in-order): {print_in_order(root)}")
```

Алгоритм корректен, потому что мы обрабатываем каждый узел ровно один раз и меняем ссылки на месте (in-place).

Сложность: O(n) времени (посещаем каждый узел ровно один раз).
